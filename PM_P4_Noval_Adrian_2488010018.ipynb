```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

```python
print("Jumlah nilai hilang per kolom:")
print(df.isna().sum())
print("\nInfo tipe data:")
df.info()
```

### Langkah 1 — Menangani nilai hilang (imputasi median/modus).

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())

print("Jumlah nilai hilang setelah imputasi:")
print(df.isna().sum())
df
```

### Langkah 2 — Memisahkan fitur (X) dan label (y).

```python
X = df.drop(columns=['membeli'])
y = df['membeli']

print("Fitur (X):")
display(X)
print("\nLabel (y):")
print(y)
```

### Langkah 3 — Encoding kategorikal (ordinal & one-hot).

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2})  # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int)                 # nominal
y = y.map({'Tidak':0,'Ya':1})

print("X setelah encoding:")
display(X)
print("\ny setelah encoding:")
print(y)
```

### Langkah 4 — Membagi data (SEBELUM scaling).

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)

print("Ukuran X_train:", X_train.shape)
print("Ukuran X_test :", X_test.shape)
display(X_train)
```

### Langkah 5 — Penskalaan (fit pada latih, transform pada latih & uji).

```python
from sklearn.preprocessing import StandardScaler

num = ['usia', 'pendapatan', 'pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

print("X_train setelah scaling:")
display(X_train)
print("\nX_test setelah scaling:")
display(X_test)
```

### Langkah 6 — Verifikasi: tidak ada nilai hilang, semua numerik, skala seragam.

```python
print("Nilai hilang di X_train:\n", X_train.isna().sum())
print("\nNilai hilang di X_test:\n", X_test.isna().sum())
print("\nTipe data X_train:\n", X_train.dtypes)
print("\nStatistik kolom numerik X_train (mean seharusnya ~0, std ~1):")
display(X_train[num].describe())
```

### 4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

`fit_transform` menghitung parameter penskalaan (mean & std untuk `StandardScaler`, atau min & max untuk `MinMaxScaler`) dari data latih, lalu langsung menerapkannya. Data uji hanya di-`transform` (tanpa `fit` ulang) agar parameter yang dipakai tetap konsisten dengan yang dipelajari dari data latih.

Jika data uji ikut di-`fit`, maka informasi dari data uji (yang seharusnya mensimulasikan data baru/tidak diketahui) akan "bocor" ke proses penskalaan, inilah yang disebut **data leakage**. Data leakage membuat evaluasi model menjadi terlalu optimis karena model secara tidak langsung sudah "mengintip" karakteristik data uji sebelum benar-benar dievaluasi.

### 1. Mengapa urutan 'split dulu, baru scaling' penting?

Jika scaling dilakukan sebelum split, statistik (mean, std, min, max) yang dipakai untuk penskalaan akan dihitung dari seluruh data, termasuk data yang seharusnya menjadi data uji. Akibatnya, informasi data uji ikut memengaruhi proses transformasi data latih, sehingga terjadi data leakage. Dengan split terlebih dahulu, data uji benar-benar merepresentasikan data baru yang belum pernah "dilihat" model maupun proses preprocessing, sehingga evaluasi performa model menjadi lebih jujur dan realistis.

### 2. Kapan label encoding dan kapan one-hot encoding?

- **Label/ordinal encoding** digunakan ketika kategori memiliki **urutan atau tingkatan yang bermakna**, misalnya `pendidikan` (SMA < S1 < S2). Urutan angka hasil encoding merepresentasikan urutan tingkatan tersebut.
- **One-hot encoding** digunakan ketika kategori bersifat **nominal**, tanpa urutan, misalnya `kota` (Bandung, Jakarta, Surabaya). Jika kategori nominal dipaksakan memakai label encoding, model bisa salah menafsirkan adanya hubungan "lebih besar/lebih kecil" antar kategori yang sebenarnya tidak ada.

### 3. Apa perbedaan normalisasi dan standardisasi?
- **Normalisasi** (mis. `MinMaxScaler`) mengubah skala fitur ke rentang tertentu, umumnya [0, 1], berdasarkan nilai minimum dan maksimum data. Cocok saat distribusi data tidak diasumsikan normal dan rentang nilai perlu dibatasi.
- **Standardisasi** (mis. `StandardScaler`) mengubah data agar memiliki mean = 0 dan standar deviasi = 1, berdasarkan distribusi data (mean & std), tanpa membatasi rentang nilai secara eksplisit. Lebih cocok digunakan saat data mendekati distribusi normal atau saat algoritma sensitif terhadap distribusi (mis. regresi linear, SVM, PCA).
